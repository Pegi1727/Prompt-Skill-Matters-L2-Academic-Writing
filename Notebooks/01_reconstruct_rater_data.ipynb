# Reconstruct Rater Data
این نوت‌بوک داده‌های خام ارزیابان را پردازش کرده و صحت محاسبات را تأیید می‌کند.

```python
import pandas as pd
import numpy as np

# Load data
df = pd.read_csv('raw_rater_scoring.csv')

# Verification logic
rater_cols = [col for col in df.columns if col.startswith('R')]
df['Calculated_Sum'] = df[rater_cols].sum(axis=1)
df['Calculated_Raw_Mean'] = df['Calculated_Sum'] / 8.0

print('Data check successful:', np.allclose(df['Calculated_Raw_Mean'], df['Raw_PSI_Mean']))
df.head()
```
